```python
import sys
import os
import torch
import pandas as pd
from tqdm import tqdm

# 1. Rutas absolutas
BASE_DIR = os.path.abspath("..")
sys.path.append(BASE_DIR)

from models.esm2_embedder import ESM2Embedder

print("1. Cargando datos...")
data_path = os.path.join(BASE_DIR, "data", "processed_dataset.csv")
df = pd.read_csv(data_path) 
secuencias = df["seq"].astype(str).tolist()

print(f"2. Cargando ESM-2 (Procesando {len(secuencias)} secuencias)...")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# Usamos el embedder de la carpeta models
embedder = ESM2Embedder(model_name="esm2_t12_35M_UR50D", device=device, batch_size=16)

print("3. Generando embeddings...")
embeddings = embedder.embed(secuencias)

print("4. Guardando...")
output_path = os.path.join(BASE_DIR, "outputs", "embeddings.pt")
os.makedirs(os.path.dirname(output_path), exist_ok=True)
torch.save(embeddings, output_path)

print(f"✅ ¡Éxito! Archivo guardado en {output_path}")
```
